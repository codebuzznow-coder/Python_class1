```python
# ============================================================
# VERSION 1: WITHOUT PROPERTIES (No Control)
# ============================================================

class ATM_NoValidation:
    def __init__(self, balance):
        self.balance = balance  # Direct access

atm1 = ATM_NoValidation(10000)

# Reading - Direct access
print(atm1.balance)  # 10000

# Writing - Direct access (NO VALIDATION!)
atm1.balance = -5000  # Negative allowed
print(atm1.balance)  # -5000

atm1.balance = "#######"  # String allowed
print(atm1.balance)  # "money"


# ============================================================
# VERSION 2: OLD METHOD WAY (Works but Ugly Syntax)
# ============================================================

class ATM_OldWay:
    def __init__(self, balance):
        self._balance = balance

    def get_balance(self):  # GETTER method
        return self._balance

    def set_balance(self, amount):  # SETTER method
        if not isinstance(amount, (int, float)):
            print("Error: Must be a number!")
            return
        if amount < 0:
            print("Error: Can't be negative!")
            return
        self._balance = amount

atm2 = ATM_OldWay(10000)

# Reading - Must call method with ()
print(atm2.get_balance())  # 10000
#         ↑↑↑↑↑↑↑↑↑↑↑ Must use () - UGLY!

# Writing - Must call method with ()
atm2.set_balance(5000)  # Works
#    ↑↑↑↑↑↑↑↑↑↑↑ Must use () - UGLY!

atm2.set_balance(-1000)  # Blocked!
atm2.set_balance("money")  # Blocked!

print(atm2.get_balance())  # 5000


# ============================================================
# VERSION 3: WITH @property (Clean and Protected)
# ============================================================

class ATM:
    def __init__(self, balance):
        self._balance = 0 ## private
        self.balance = balance  # Uses setter

    @property
    def balance(self):  # GETTER
        return self._balance

    @balance.setter
    def balance(self, amount):  # SETTER
        if not isinstance(amount, (int, float)):
            print("Error: Must be a number!")
            return
        if amount < 0:
            print("Error: Can't be negative!")
            return
        self._balance = amount

atm3 = ATM(10000)

# Reading - NO (), looks like variable
print(atm3.balance)  # 10000
#         ↑↑↑↑↑↑↑ NO () - CLEAN!

# Writing - NO (), looks like assignment
atm3.balance = 5000  # Works
#    ↑↑↑↑↑↑↑ NO () - CLEAN!

atm3.balance = -1000  # Blocked!
atm3.balance = "money"  # Blocked!

print(atm3.balance)  # 5000




# HOW TO READ:
value = atm1.balance           # Direct (no protection)
value = atm2.get_balance()     # Method (needs parentheses)
value = atm3.balance           # Property (clean, protected)

# HOW TO WRITE:
atm1.balance = 5000            # Direct (no protection)
atm2.set_balance(5000)         # Method (needs parentheses)
atm3.balance = 5000            # Property (clean, protected)
```
